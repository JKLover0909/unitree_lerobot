```python
from unitree_sdk2py.core.channel import ChannelSubscriber, ChannelFactoryInitialize
from inspire_sdkpy import inspire_dds
import threading
import numpy as np

class TactileHandler:
    def __init__(self, LR='l'):
        # ChannelFactoryInitialize(0)
        
        # Subscribe to tactile topic
        self.sub_touch = ChannelSubscriber("rt/inspire_hand/touch/" + LR, inspire_dds.inspire_hand_touch)
        self.sub_touch.Init(self.update_data_touch, 10)
        
        self.touch = {}
        self.data_touch_lock = threading.Lock()
    
    def update_data_touch(self, msg: inspire_dds.inspire_hand_touch):
        """Callback khi nhận dữ liệu tactile mới"""
        with self.data_touch_lock:
            # Danh sách các sensor trong inspire_hand_touch message
            sensors = [
                'fingerone_tip_touch',      # Pinky tip (9 elements)
                'fingerone_top_touch',      # Pinky top (96 elements)
                'fingerone_palm_touch',     # Pinky palm (80 elements)
                'fingertwo_tip_touch',      # Ring tip (9 elements)
                'fingertwo_top_touch',      # Ring top (96 elements)
                'fingertwo_palm_touch',     # Ring palm (80 elements)
                'fingerthree_tip_touch',    # Middle tip (9 elements)
                'fingerthree_top_touch',    # Middle top (96 elements)
                'fingerthree_palm_touch',   # Middle palm (80 elements)
                'fingerfour_tip_touch',     # Index tip (9 elements)
                'fingerfour_top_touch',     # Index top (96 elements)
                'fingerfour_palm_touch',    # Index palm (80 elements)
                'fingerfive_tip_touch',     # Thumb tip (9 elements)
                'fingerfive_top_touch',     # Thumb top (96 elements)
                'fingerfive_middle_touch',  # Thumb middle (80 elements)
                'fingerfive_palm_touch',    # Thumb palm (80 elements)
                'palm_touch'                # Palm (240 elements)
            ]
            
            for sensor_name in sensors:
                value = getattr(msg, sensor_name, None)
                if value is not None:
                    self.touch[sensor_name] = np.array(value)
    
    def get_tactile_data(self):
        """Lấy dữ liệu tactile hiện tại"""
        with self.data_touch_lock:
            return self.touch.copy()

# Sử dụng
handler = TactileHandler(LR='l')  # 'l' cho left hand, 'r' cho right hand

import time
for i in range(10):
    tactile = handler.get_tactile_data()
    if tactile:
        print(f"\n=== Lần đọc {i+1} ===")
        for sensor_name, data in tactile.items():
            print(f"{sensor_name}: shape={data.shape}, min={data.min()}, max={data.max()}, mean={data.mean():.2f}")
    time.sleep(0.5)
```

```python
# Cell 1: Import
import time
import sys
from unitree_sdk2py.core.channel import ChannelPublisher, ChannelSubscriber, ChannelFactoryInitialize
from inspire_sdkpy import inspire_hand_defaut, inspire_dds, inspire_sdk
import numpy as np
import threading

# Cell 2: Khởi tạo DDS
# ChannelFactoryInitialize(0)

# Cell 3: Tạo Publisher để điều khiển
publ = ChannelPublisher("rt/inspire_hand/ctrl/l", inspire_dds.inspire_hand_ctrl)
publ.Init()

# Cell 4: Gửi lệnh điều khiển (giống code cũ)
cmd = inspire_hand_defaut.get_inspire_hand_ctrl()
cmd.mode = 0b1101
cmd.angle_set = [300, 300, 300, 300, 1000, 1000]
cmd.force_set = [200] * 6
cmd.speed_set = [350] * 6
publ.Write(cmd)

# Cell 5: ĐỌC TACTILE DATA - CÁCH ĐÚNG
class TactileReader:
    def __init__(self, LR='l'):
        self.touch = {}
        self.lock = threading.Lock()
        self.sub = ChannelSubscriber("rt/inspire_hand/touch/" + LR, inspire_dds.inspire_hand_touch)
        self.sub.Init(self._callback, 10)
    
    def _callback(self, msg: inspire_dds.inspire_hand_touch):
        with self.lock:
            # Lưu tất cả tactile data
            self.touch = {
                'pinky_tip': np.array(msg.fingerone_tip_touch),
                'pinky_top': np.array(msg.fingerone_top_touch),
                'pinky_palm': np.array(msg.fingerone_palm_touch),
                'ring_tip': np.array(msg.fingertwo_tip_touch),
                'ring_top': np.array(msg.fingertwo_top_touch),
                'ring_palm': np.array(msg.fingertwo_palm_touch),
                'middle_tip': np.array(msg.fingerthree_tip_touch),
                'middle_top': np.array(msg.fingerthree_top_touch),
                'middle_palm': np.array(msg.fingerthree_palm_touch),
                'index_tip': np.array(msg.fingerfour_tip_touch),
                'index_top': np.array(msg.fingerfour_top_touch),
                'index_palm': np.array(msg.fingerfour_palm_touch),
                'thumb_tip': np.array(msg.fingerfive_tip_touch),
                'thumb_top': np.array(msg.fingerfive_top_touch),
                'thumb_middle': np.array(msg.fingerfive_middle_touch),
                'thumb_palm': np.array(msg.fingerfive_palm_touch),
                'palm': np.array(msg.palm_touch)
            }
    
    def get_data(self):
        with self.lock:
            return self.touch.copy()

# Tạo reader
tactile_reader = TactileReader(LR='l')

# Đợi 1 giây để nhận dữ liệu đầu tiên
time.sleep(1.0)

# Đọc tactile trong 10 giây
print("Bắt đầu đọc tactile data...\n")
start_time = time.time()

try:
    while time.time() - start_time < 10.0:
        tactile = tactile_reader.get_data()
        
        if tactile:
            print(f"[{time.time() - start_time:.2f}s]")
            for name, data in tactile.items():
                if len(data) > 0:
                    print(f"  {name}: shape={data.shape}, min={data.min()}, max={data.max()}, mean={data.mean():.1f}")
            print()
        else:
            print(f"[{time.time() - start_time:.2f}s] Đang chờ dữ liệu...")
        
        time.sleep(0.5)
        
except KeyboardInterrupt:
    print("\n⚠️ Đã dừng")

print("✅ Hoàn thành")
```
